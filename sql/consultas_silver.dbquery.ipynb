```sql

-- ============================================================
-- CONSULTA 1: TOTAL DE CLUBES DE FÚTBOL
-- Objetivo: contar los clubes registrados en la tabla Silver.
-- ============================================================

SELECT
    COUNT(*) AS total_football_clubs
FROM big_data_team3.fabrizio.clubes_futbol_silver;


-- ============================================================
-- CONSULTA 2: EVENTOS INTERNACIONALES POR PAÍS
-- Objetivo: identificar los países con más eventos registrados.
-- ============================================================

SELECT
    pais_evento AS country,
    COUNT(*) AS total_events
FROM big_data_team3.fabrizio.eventos_internacionales_silver
WHERE pais_evento IS NOT NULL
GROUP BY pais_evento
ORDER BY total_events DESC;


-- ============================================================
-- CONSULTA 3: EVENTOS INTERNACIONALES POR FEDERACIÓN
-- Objetivo: identificar las federaciones con más eventos.
-- ============================================================

SELECT
    federacion AS federation,
    COUNT(*) AS total_events
FROM big_data_team3.fabrizio.eventos_internacionales_silver
WHERE federacion IS NOT NULL
GROUP BY federacion
ORDER BY total_events DESC;


-- ============================================================
-- CONSULTA 4: EVENTOS INTERNACIONALES POR AÑO
-- Objetivo: analizar la cantidad de eventos por año.
-- ============================================================

SELECT
    YEAR(fecha_inicio) AS event_year,
    COUNT(*) AS total_events
FROM big_data_team3.fabrizio.eventos_internacionales_silver
WHERE fecha_inicio IS NOT NULL
GROUP BY YEAR(fecha_inicio)
ORDER BY event_year ASC;


-- ============================================================
-- CONSULTA 5A: TOTAL DE REGISTROS EN LAS SEIS TABLAS SILVER
-- Objetivo: comparar el volumen de datos de cada tabla.
-- ============================================================

SELECT 'pad_silver' AS table_name, COUNT(*) AS total_rows
FROM big_data_team3.fabrizio.pad_silver

UNION ALL

SELECT 'eventos_internacionales_silver', COUNT(*)
FROM big_data_team3.fabrizio.eventos_internacionales_silver

UNION ALL

SELECT 'clubes_futbol_silver', COUNT(*)
FROM big_data_team3.fabrizio.clubes_futbol_silver

UNION ALL

SELECT 'clubes_renade_silver', COUNT(*)
FROM big_data_team3.fabrizio.clubes_renade_silver

UNION ALL

SELECT 'decan_silver', COUNT(*)
FROM big_data_team3.fabrizio.decan_silver

UNION ALL

SELECT 'ligas_renade_silver', COUNT(*)
FROM big_data_team3.fabrizio.ligas_renade_silver;


-- ============================================================
-- CONSULTA 5B: VALORES NULOS EN EVENTOS INTERNACIONALES
-- Objetivo: identificar columnas con datos faltantes.
-- ============================================================

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN fecha_corte IS NULL THEN 1 ELSE 0 END)
        AS fecha_corte_nulls,
    SUM(CASE WHEN item IS NULL THEN 1 ELSE 0 END)
        AS item_nulls,
    SUM(CASE WHEN federacion IS NULL THEN 1 ELSE 0 END)
        AS federacion_nulls,
    SUM(CASE WHEN evento IS NULL THEN 1 ELSE 0 END)
        AS evento_nulls,
    SUM(CASE WHEN pais_evento IS NULL THEN 1 ELSE 0 END)
        AS pais_evento_nulls,
    SUM(CASE WHEN ciudad_evento IS NULL THEN 1 ELSE 0 END)
        AS ciudad_evento_nulls,
    SUM(CASE WHEN fecha_inicio IS NULL THEN 1 ELSE 0 END)
        AS fecha_inicio_nulls,
    SUM(CASE WHEN fecha_fin IS NULL THEN 1 ELSE 0 END)
        AS fecha_fin_nulls,
    SUM(CASE WHEN deportista IS NULL THEN 1 ELSE 0 END)
        AS deportista_nulls,
    SUM(CASE WHEN puesto IS NULL THEN 1 ELSE 0 END)
        AS puesto_nulls,
    SUM(CASE WHEN especialidad IS NULL THEN 1 ELSE 0 END)
        AS especialidad_nulls,
    SUM(CASE WHEN _rescued_data IS NULL THEN 1 ELSE 0 END)
        AS rescued_data_nulls
FROM big_data_team3.fabrizio.eventos_internacionales_silver;







```
